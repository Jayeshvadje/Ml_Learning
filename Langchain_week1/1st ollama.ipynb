[Ollama](https://ollama.com/) is an open-source framework and runtime application designed to let you download, manage, and run Large Language Models (LLMs) locally on your own computer. It acts as a lightweight package manager and server for AI models, allowing developers and tech enthusiasts to use powerful AI capabilities completely offline. [1, 2]
## Core Benefits

* Data Privacy: Because the model executes entirely on your hardware, your prompts and data never touch a third-party cloud server.
* Cost Efficiency: You avoid per-token billing, monthly API fees, and subscription costs associated with commercial AI providers.
* Offline Access: Once a model is fully downloaded, you can interact with it without an active internet connection. [2, 3, 4, 5]

## Key Features

* Single-Command CLI: Downloading and starting a chat with an optimized model is as simple as typing ollama run <model-name> into your terminal. [1, 6]
* Local REST API: It automatically starts an inference server on localhost:11434, serving standard API endpoints that let developers connect apps or frameworks like LangChain directly to the local model. [1]
* Custom Modelfiles: Similar to Docker, Ollama uses a Modelfile configuration that allows you to easily adjust system prompts, temperatures, and parameters to build custom AI personas. [1, 5]
* Wide Compatibility: It natively supports macOS, Linux, and Windows, optimizing performance automatically across both CPUs and dedicated GPUs (NVIDIA/AMD). [1, 2, 7]

## Supported Models
Ollama hosts a massive [library of open-source models](https://ollama.com/library) spanning various sizes and specialized use cases: [8, 9, 10, 11]

* General Text & Logic: Popular weights like Meta's Llama series, Mistral, Google's Gemma, and Alibaba's Qwen.
* Reasoning Models: Specialized models capable of step-by-step chain-of-thought processing, such as DeepSeek.
* Multimodal: Options capable of parsing and analyzing visual images alongside text.
* Code & Embedding: Models explicitly tuned for coding assistance (like Dolphin-Mistral) or generating text embeddings for data indexing. [1, 2, 8, 12, 13]

## Growth and Ecosystem Extension
While Ollama natively provides a command-line terminal interface, a massive community ecosystem has formed around it. Users frequently pair it with graphic wrappers like Open WebUI to create a local web interface that functions exactly like a private ChatGPT clone. Additionally, [Ollama's documentation](https://docs.ollama.com/cloud) outlines hybrid options like Cloud Models, which allow resource-constrained computers to offload massive workloads seamlessly to remote infrastructure when required. [1, 14, 15, 16]



```python
from langchain_ollama import OllamaLLM


ollama = OllamaLLM(
    base_url="http://localhost:11434",
    model="mario"
)

# 3. Use .invoke() with a quoted string prompt
response = ollama.invoke("who are you?")
print(response)
```

```python
response = str(input())
response=ollama.invoke(response)
print(response)
```
